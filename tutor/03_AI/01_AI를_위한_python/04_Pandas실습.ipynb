```python
# 라이브러리 및 환경 설정
import numpy as np
import pandas as pd

# 항상 같은 무작위 결과를 위해 시드 고정
np.random.seed(42)
```

```python
# 50명의 가상 사용자 데이터 생성
N = 50
data = {
    '사용자_ID': range(100, 100 + N),
    '나이': np.random.randint(20, 50, N),
    '구매액': np.random.randint(50000, 300000, N),
    '지역': np.random.choice(['서울', '부산', '대구', '인천'], N),
    '평가점수': np.random.randn(N) * 10 + 70 # 평균 70점, 정규분포
}

df = pd.DataFrame(data)

# 임의로 결측치(NaN)를 주입 (약 10% 정도)
    # sample: 무작위 샘플링
    # frac: 비율, random_state: 시드
    # .index: 인덱스 추출
    # loc: 인덱스 기반 접근
        # loc[행, 열]
df.loc[df.sample(frac=0.10, random_state=42).index, '구매액'] = np.nan
df.loc[df.sample(frac=0.05, random_state=42).index, '평가점수'] = np.nan
```
