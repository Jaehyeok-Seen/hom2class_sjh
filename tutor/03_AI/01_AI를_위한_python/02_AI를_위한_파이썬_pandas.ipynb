### 3-1-1. DataFrame

- Pandas의 핵심. NumPy를 활용한 빠른 연산과 더불어 `namespace`도 함께 있어 직관적으로 데이터를 다룰 수 있음.
1. 데이터 초기화
    - 파이썬의 `딕셔너리`나 `리스트`의 데이터를 DataFrame으로 변환 가능

```python
import pandas as pd

# 리스트 데이터 초기화
data_list = [
    ['Alice', 25, 'New York'],
    ['Bob', 30, 'Los Angeles'],
    ['Charlie', 35, 'Chicago']
]

df_list = pd.DataFrame(data_list, columns=['이름', '나이', '도시'])
print(df_list)
```

```python
# 딕셔너리 데이터 초기화
data = {
    '이름': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', None, 'Grace', 'Hannah', 'Ian', 'Jack'],
    '나이': [25, 30, 35, None, 28, 22, None, 29, 31, 27],
    '도시': ['New York', None, 'Chicago', 'Houston', 'Phoenix', 'Philadelphia', 'San Antonio', None, 'San Diego', 'Dallas']
}

df = pd.DataFrame(data)
print(df)
```

### 3-1-2. 데이터 탐색과 선택

- 데이터 기본 정보 조회
    1. `.head()`/ `.tail()`: 상위 5개 / 하위 5개 행만 조회
    2. `.info()`: 각 컬럼이 어떤 종료의 데이터를 담고 있는지 (숫자인지, 문자인지) 비어있는 값(결측치)은 없는지 요약 정보를 알려줌.
    3. `.describe()`: 숫자형 컬럼들의 통계 정보를 요약해서 보여줌. 평균, 최댓값, 최솟값 등

```python
print('\n 상위 5개 행 조회')
print(df.head())
```

```python
print('\n 하위 5개 행 조회')
print(df.tail())
```

```python
print('\n 데이터 정보 조회')
print(df.info())
```

```python
print('\n 통계 정보 조회')
print(df.describe())
```

- 데이터 상세 조회와 필터링
  1. **컬럼 선택:** 특정 컬럼을 선택하여, 해당 컬럼에 대한 데이터만 반환. 이때 반환되는 데이터 타입은 `Series` 라는 1차원 데이터가 됨

```python
print('\n 특정 열(나이) 조회')
age_series = df['나이']
print(age_series)
print(type(age_series))
```

```python
print('\n 여러 열(이름, 도시) 조회')
name_city_df = df[['이름', '도시']]
print(name_city_df)
print(type(name_city_df))
```

  2. **행 선택**
      1. **`.loc` (Label-based):** `이름(Label)`을 기준으로 행을 선택. 
      2. **`.iloc` (Integer-based):** `순서(index)`를 기준으로 행을 선택

```python
print('\n 특정 행(이름이 Bob인 행) 조회 using .loc')
bob_row = df.loc[df['이름'] == 'Bob']
print(bob_row)
print(type(bob_row))
```

```python
print('\n 특정 행(두 번째 행) 조회 using .iloc')
second_row = df.iloc[1]
print(second_row)
print(type(second_row))
```

3. **조건 필터링**
    - **불리언 인덱싱 (Boolean Indexing)**
      - 조건 만들기: 원하는 조건을 만들면, 해당 조건에 대한 True or False로 채워진 리스트가 반환
      - 필터 적용: 이 True/False 리스트를 다시 DataFrame에 씌우면, True인 행만 반환

```python
print('\n 나이가 30 이상인 조건 필터링')
age_filter = df['나이'] >= 30
print(age_filter)
```

```python
print('\n 나이가 30 이상인 행 조회')
filtered_df = df[age_filter]
print(filtered_df)

# print('\n 나이가 30 이상인 행을 한 코드로')
# print(df[df['나이'] >= 30])
```

```python
print('\n 나이가 30 이상이고 도시가 Chicago인 행 조회')
print(df[(df['나이'] >= 30) & (df['도시'] == 'Chicago')])
```

### 3-1-3. 데이터 가공 및 분석

- 결측치 조회 및 가공
  1. `.isnull()`: 값이 비어 있는 경우 True를 반환
  2. `.fillna()`: 값이 비어 있는 경우, 지정한 값으로 빈칸을 채움 (일반적으로 평균값, 중앙값, 0)
  3. `.dropna()`: 결측치가 있는 행 삭제

```python
# 결측치 조회
print('\n 결측치 조회')
print(df.isnull())
```

```python
# 결측치가 있는 행 조회
print('\n 결측치가 있는 행 조회')
print(df[df.isnull().any(axis=1)])
```

```python
# 결측치 채우기
print('\n 나이의 결측치를 평균값으로 채우기')
mean_age = df['나이'].mean()
print(f'나이 평균값: {mean_age}')

df['나이'] = df['나이'].fillna(mean_age)
print(df)
```

```python
# 결측치가 있는 행 삭제
print('\n 결측치가 있는 행 삭제')
df_dropped = df.dropna()
print(df_dropped)
```

- 그룹화 및 통계
    1. **`.groupby()`:** 데이터를 특정 컬럼과 같은 범주로 묶음
    2. **aggregation(집계):** 묶인 그룹에 대해 `.mean()`, `.sum()`, `.count()`등을 적용하여 통계 결과를 얻음
    3. **정렬**

```python
# 그룹화
grouped = df.groupby('도시')
print(grouped)  # 그룹화 객체 출력
```

```python
print('\n 도시별 나이 평균')
print(grouped['나이'].mean())
```

```python
# 집계
print('\n 도시별 나이 최대값과 최소값')
print(grouped['나이'].agg(['max', 'min']))
```

```python
# aggregate: 여러 집계 함수를 동시에 적용
# 도시별 나이 최대값과 이름 개수
# grouped.aggregate({'나이': 'max', '이름': 'count'})
```

```python
# 정렬
print('\n 나이 기준 내림차순 정렬')
sorted_df = df.sort_values(by='나이', ascending=False)
print(sorted_df)
```
