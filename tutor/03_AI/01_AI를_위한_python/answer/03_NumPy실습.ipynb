```python
# 라이브러리 및 환경 설정
import numpy as np

# 항상 같은 무작위 결과를 위해 시드 고정
np.random.seed(42)
```

### 1. 배열 초기화 및 속성 파악

```python
# 5행 3열의 크기를 가지는 0부터 1 사이의 무작위 값으로 채워진 2차원 배열 X를 생성하세요.
X = np.random.random(size=(5, 3))

# 생성된 배열의 shape, size, ndim 속성을 출력하세요.
print("X의 모양:", X.shape)
print("X의 총 요소 수:", X.size)
print("X의 차원:", X.ndim)
print(X)
```

### 2. 브로드캐스팅 응용

```python
# 1. 5x5 크기의 무작위 배열 M에 10을 더하는 연산을 수행하세요.
M = np.random.rand(5, 5)
M_scaled = M + 10

print(M)
print(M_scaled)
```

```python
# 2. 브로드캐스팅을 이용하여 2차원 배열 M의 각 열(Column)에 대해 다른 값(벡터 V)을 더하세요.
V = np.array([100, 200, 300, 400, 500])
# V를 (5, 1) 형태로 만들어 (5, 5) 배열 M에 브로드캐스팅 되도록 연산하세요.

# 브로드캐스팅을 이용한 덧셈
# reshape를 사용하여 V의 형태를 (5, 1)로 변경
    # -1은 자동으로 크기를 계산하도록 함
    # 1은 열의 개수를 의미
    # 따라서 V.reshape(-1, 1)은 V를 (5, 1) 형태로 변환
M_broad = M + V.reshape(-1, 1)
print("\nM + V (브로드캐스팅 결과):\n", M_broad)
```

### 3. 데이터 정규화

```python
# 100x2 크기의 평균 10, 분산 25인 데이터를 생성하세요.
    # 평균 10, 분산 25인 정규분포를 따르는 데이터를 생성하기 위해
    # np.random.randn() 함수를 사용하여 표준 정규분포(평균 0, 분산 1)를 따르는 데이터를 생성한 후,
    # 이를 원하는 평균과 분산으로 변환합니다.
    # 변환 공식: X_new = X * std + mean
    # 여기서 std는 표준편차(분산의 제곱근), mean은 원하는 평균입니다.
D = np.random.randn(100, 2) * 5 + 10 # 평균 10, 분산 25인 데이터
```

```python
# D의 평균과 표준편차를 계산하세요.
answer_1 = D.mean()
answer_2 = D.std()

print(f'D의 평균: {answer_1}')
print(f'D의 표준편차: {answer_2}')
```

```python
# 위에서 계산한 D의 평균과 표준편차를 사용하여 D의 각 요소를 정규화합니다.
    # 정규화 공식: Z = (X - mean) / std
Z = (D - answer_1) / answer_2
print(Z)
```

### 4. 행렬 곱과 전치 응용

```python
# 입력 데이터 X_input (10, 3)과 가중치 W (3, 1) 행렬을 생성하세요.
X_input = np.random.rand(10, 3)
W = np.random.rand(3, 1)

print(X_input)
print(W)
```

```python
# X_input과 가중치 W를 행렬 곱하여 예측 결과(Y_pred)를 계산하세요.
Y_pred = X_input @ W

print(Y_pred)
```

```python
# X_input과 W를 전치하여 행렬곱을 수행하세요.
# (1, 10) 배열이 나오도록 연산하세요
Y_dot = W.T @ X_input.T
print("\n내적 연산 결과의 모양:", Y_dot.shape)
```

### extra. 역행렬 계산

```python

# 정방 행렬 A를 생성하고, np.linalg.inv()를 사용하여 역행렬(A_inv)을 계산하세요.
# A @ A_inv 의 결과가 단위행렬(np.eye)과 같은지 확인하세요. (최적해 계산의 원리)
A = np.array([[4, 5], [2, 6]])
print(A)

# 역행렬 계산
    # np.linalg.inv() 함수는 정방 행렬의 역행렬을 계산
    # 역행렬은 원래 행렬과 곱했을 때 단위 행렬이 되는 행렬
A_inv = np.linalg.inv(A)

# A와 역행렬의 곱셈 
    # 최적화 문제에서 해를 구할 때, 역행렬을 이용하여 시스템의 해를 구할 수 있음
    # 단위행렬은 행렬 곱셈의 항등원 역할을 하기 때문에, 
    # A @ A_inv가 단위행렬이 되면 A_inv가 A의 진정한 역행렬임을 의미
I_check = A_inv @ A

print("\n 역행렬:\n", A_inv)
print("\n A @ A_inv (단위행렬 확인):\n", I_check)
print("\n 단위행렬:\n", np.eye(2, 2))
```
