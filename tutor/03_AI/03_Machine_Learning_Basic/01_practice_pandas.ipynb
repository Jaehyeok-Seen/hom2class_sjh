### 8. Color intensity가 10 이상인 샘플의 비율(%)을 구해서 high_color_ratio 변수에 담아주세요!
- pandas의 대괄호 `[]` 는 파이썬의 딕셔너리와 리스트의 대괄호 접근법과는 다른 기능을 제공
    - 기본적으로 파이썬 데이터 타입과 같이 `컬럼` 을 선택할 때도 사용
        - ex) `df.['quality'] = y` 로 y의 값을 quality 열에 할당 하거나 조회 할 때 사용
    - 행을 필터링 할 때
        - `df[df['color_intensity'] >= 10]`처럼 대괄호 안에 `True`/`False`로 이루어진 `Series`(불리언 마스크)를 넣으면, `pandas`는 `True`에 해당하는 행들만 골라서 할당합니다.

````python
# 먼저 조건에 맞는 샘플들을 골라내기 위해 불리언 시리즈를 만듭니다.
filtered_df = df['color_intensity'] >= 10
# print(filtered_df)
# 조건에 맞는 샘플들만 따로 빼냅니다.
high_color_samples = df[filtered_df]
# print(high_color_samples)

# 그 샘플들의 개수를 전체 샘플 수로 나눈 뒤, 100을 곱해 비율(%)을 구합니다.
high_color_ratio = len(high_color_samples) / len(df) * 100

print(f"Color intensity가 10 이상인 샘플 비율: {high_color_ratio:.2f}%") # 3.93%
````

### 9. Ash 특성에서 최소값을 가진 샘플의 클래스를 구해서 min_ash_class 변수에 담아주세요!

- `idxmax()`, `idxmin()`: 가장 크거나 작은 값의 인덱스를 찾는 법
- 위치(location) 기반으로 원하는 값을 찾는 방법은? `loc()`
    1. `df.loc['행이름', '열이름']` : 하나의 값 찾기
    2. `df.loc['행이름']` : 하나의 행 전체 찾기
    3. `df.loc[:, '열이름']` : 하나의 열 전체 찾기
- pandas 데이터프레임의 기본 인덱스는 `번호(정수 인덱스)`  == `0, 1, 2, ...`
    - 사용자 지정 인덱스도 사용 가능
        - 예를들어, 와인 이름, 상품 코드와 같은 고유한 열이 있다면, 해당 열을 `이름` 으로 지정
        - 즉, `인덱스로 특정 컬럼을 지정` 할 수 있음. (`df.set_index(’와인 이름’)`
- 만약, 이런 지정 인덱스를 사용 중이지만, 정수 형태로 loc을 쓰고 싶다면 `df.iloc()` 을 활용 가능
    - `iloc(행 번호, 열 번호)` : 사용 방법은 loc과 동일

````python
# 최소값을 가진 샘플의 인덱스를 먼저 찾습니다.
min_ash_index = df['ash'].idxmin()
# max_ash_index = df['ash'].idxmax()
# print(min_ash_index, max_ash_index)

# 그 인덱스를 사용해 해당 샘플의 클래스를 구합니다.
min_ash_class = df.loc[min_ash_index, 'quality']

print(f"Ash 최소값을 가진 샘플의 클래스: {min_ash_class}")  # 1
````

- Q. 상관 관계처럼, 정량적 평가를 하고 싶습니다.
    - A. 코드 참고

````python
# `pd.cut()`의 결과가 의미하는 것: 원래 `prolone` 데이터가 어떤 구간에 속하는지를 나타냄
# `pd.cut(기준 열, 구간의 개수 (bin)` : 기준 열의 값을 10개의 구간이 되도록 알아서 나눔
    # 그 결과를 `(시작 값, 끝 값]` 형태의 구간으로 만들어 줌.
    # `binned_proline` 출력 결과는 각 행 (0번 와인, 1번 와인…)이 어느 구간 `(시작 값, 끝 값]` 에 속하는지를 보여줌
binned_proline = pd.cut(df['proline'], bins=10)
# print(binned_proline)

# 등급별로 그룹을 나누고, 각 구간에 몇 개씩 있는지 셉니다.
proline_by_class = binned_proline.groupby(df['quality']).value_counts()
# 구성: level_0(quality), level_1(구간) -> 값(빈도수)
print(proline_by_class)

# level=0: quality별로 그룹화
# level=1: 구간별로 그룹화
# level=2: 빈도수별로 그룹화

# quality별로 그룹화해서, 각 클래스에서 가장 높은 빈도수를 찾습니다.
peak_counts = proline_by_class.groupby(level=0).max()
# print(peak_counts)

# 이제 가장 높은 빈도를 가진 클래스를 찾습니다.
proline_peak_class = peak_counts.idxmax()

print("Proline 분포에서 가장 높은 피크를 보이는 클래스 번호:", proline_peak_class) # 1
````

### 12. Alcohol과 가장 상관관계가 높은 특성 이름을 구해서 top_corr_with_alcohol 변수에 담아주세요!

- **상관관계(Correlation):** 두 변수가 얼마나 **선형적인 관계를 가지는지 나타내는 값**
    - **항상 -1과 1사이**
    - **1에 가까울수록:** 한 변수가 증가할 때 다른 변수도 같이 증가하는 **양의 상관관계**
    - **-1에 가까울수록:** 한 변수가 증가할 때 다른 변수는 감소하는 **음의 상관관계**
    - **0에 가까울수록:** 두 변수 사이의 관계가 거의 없다
- **상관 관계가 계산되는 과정**
    
    `피어슨 상관계수`
    
    - **$xˉ$ (x바):** x의 평균
    
    $r = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}$
    
    **1. 분자(**$∑(xi−xˉ)(yi−yˉ)$**):**
    이 부분은 두 변수가 **같은 방향으로 움직이는지**를 측정
    
    - 만약 $x$가 평균보다 클 때 $y$도 평균보다 크고, $x$가 평균보다 작을 때 $y$도 평균보다 작다면, 이 값은 **양수**
    
    **2. 분모(**$∑(xi−xˉ)2∑(yi−yˉ)2$**):**
    이 부분은 분자를 **-1과 1 사이의 값으로 만들어주는 역할**
    
    - '표준화'라고도 부르는데, 데이터의 크기나 단위에 상관없이 일정한 범위로 바꿔 줌
    
- **탐색적 데이터 분석(Exploratory Data Analysis, EDA)**
    - 목표: 데이터의 특징을 파악하고, 숨겨진패턴을 찾고, 이상한 점은 없는지 찾는 것
        - 지도 학습 과정에서 사용할 `특징` 과 `정답` 을 예측하는데 활용함.
        - 비지도 학습을 진행 할 때, EDA를 통해 데이터가 몇 개의 그룹으로 뭉쳐 있는지, 어떤 특징들이 서로 관련이 있는지 미리 파악.

### **2. 선형 회귀 모델 수립**

- **모델 가정:** 선형 회귀 모델은 예측하고자 하는 목표 변수(와인 등급)와 독립 변수(예: `Proline` 값) 사이에 선형 관계가 존재한다는 `가정`을 기반으로 함
- **모델 목표:** `y = Wx + b`라는 선형 방정식을 통해 와인 등급을 예측하는 모델을 수립하며, 이 모델의 목표는 잔차(예측값과 실제값의 차이)를 최소화하는 최적의 가중치(`W`)와 편향(`b`)을 찾는 것
