# 2. 특성 간의 관계 분석

## 2-1. 산점도와 상관관계

### **2-1-1. 산점도 (Scatter Plot)**

- 두 변수(특성)간의 관계를 파악하기 위해, 각 데이터 포인트를 2차원 평면에 **점으로 나타낸 그래프**
- 변수들이 어떤 패턴(선형, 비선형, 무작위 등)을 보이는지 시각적으로 확인

### 2-1-2. 상관관계(Correlation)

- 두 변수가 함께 변화하는 **경향성을** 나타내는 통계적 척도
- 일반적으로 -1에서 1 사이의 값을 가지며, **상관 계수(Correlation Coefficient)**로 표현
    1. **양의 상관관계 (1에 가까움):** 한 변수가 증가할 때 다른 변수도 증가하는 경향
    2. **음의 상관관계 (-1에 가까움):** 한 변수가 증가할 때 다른 변수는 감소하는 경향
    3. **상관관계 없음 (0에 가까움):** 두 변수 사이에 뚜렷한 선형 관계가 없음

## 2-2. 변수 간의 관계를 분석해야 하는 이유

- 예측 모델의 근본적인 목표는 입력 변수 (X)와 결과 변수 (Y) 사이의 수학적 관계를 규명하는 것
- 이번 수업에서는 `과제 제출률`이 `시험 점수`에 어떠한 영향을 미치는지 알아볼 것
    - 미치는 영향이 어떠한 지를 분석하여 유의미한 예측 모델을 만들기 위해 필요한 정보는 무엇인지를 알아낼 수 있음.
1. 이 과정을 위해 산점도를 통해 관계를 시각적으로 확인하고,
2. 상관 계수를 통해 그 관계의 방향성과 강도를 객관적인 수치로 제공

## 2-3. 산점도 시각화 및 상관 계수 계산

- `plt.scatter()`: Matplotlib을 사용하여 `assignment_rate`와 `final_score` 두 변수 간의 관계를 산점도로 시각화
- `df.corr()`: Pandas의 `.corr()` 메서드를 사용하여 데이터 프레임의 모든 숫자형 변수 간의 상관계수를 계산
    - **참고) 상관 관계 분석 종류 중 일부**
        1. 피어슨 상관계수 (Pearson correlation coefficient)
            - 가장 일반적으로 사용되는 상관계수
        2. 스피어만 상관계수 (Spearman's rank correlation coefficient)
            - 순위 기반 상관계수
        3. 파이 상관계수 (Phi coefficient)
            - 범주형 변수 간의 상관계수
            
### 2-3-1. 산점도를 이용한 시각적 관계 분석

- x축에는 원인이 될 수 있는 `assignment_rate(과제 제출률)`를, y축에는 결과인 `final_score`를 배치하여, 두 변수 간의 패턴을 확인

```python
# Google Drive 연결
from google.colab import drive
drive.mount('/content/drive')

base_path = '/content/drive/MyDrive/Colab Notebooks/'
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# 실습에 사용할 메인 데이터셋 로드
df_main = pd.read_csv(base_path + 'virtual_student_data.csv')

# 메인 데이터셋 정보 확인
print("Main Dataset Info:")
df_main.info()
```

```python
# 1. 분석할 두 변수(특성) 선택
assignment_rate = df_main['assignment_rate']
final_score = df_main['final_score']

# 2. 산점도 그리기
# figure: 그래프 크기 설정
    # figsize: (가로, 세로) 단위는 인치
plt.figure(figsize=(10, 6))
# scatter 함수로 산점도 생성
    # x축: assignment_rate, y축: final_score
    # s, c, marker 등 다양한 옵션 설정 가능
        # s: 점 크기(size)
        # c: 점 색상(color)
        # marker: 점 모양(marker style)
        # alpha: 점 투명도(transparency)
        # edgecolor: 점 테두리 색상
        # cmap: 색상 맵(color map)
        # linewidth: 점 테두리 두께
        # vmin, vmax: 색상 범위 설정
        # norm: 색상 정규화 방법
plt.scatter(assignment_rate, final_score, alpha=0.7) # alpha는 점의 투명도

# 3. 그래프에 제목과 라벨 추가
plt.title('Assignment Rate vs. Final Score')
plt.xlabel('Assignment Submission Rate (%)')
plt.ylabel('Final Exam Score')
plt.grid(True)
plt.show()
```

### 2-3-2. 상관 계수를 이용한 정량적 관계 분석

- `.corr()`메서드를 통해 산점도에서 확인한 시각적 패턴을 정량적인 수치로 확인
   - pandas의 corr() 함수는 기본적으로 피어슨 상관계수를 계산
        - **cov(X, Y):** X와 Y의 공분산(covariance) 
            - (공분산: 두 변수 간의 관계를 나타내는 지표)
        - **E:** 기대값 (평균)
        - $cov(X, Y) = E[(X - mean(X)) * (Y - mean(Y))]$
        - $r = cov(X, Y) / (std(X) * std(Y))$

- 단, student_id와 같은 식별자 열은 상관계수 계산에서 제외할 것.
  - 원본 df_main은 그대로 두고, 상관 계수 계산용 복사본만 별도로 담아서 사용

```python

# 복사본 생성 후 student_id 열 제거
df_main_copy = df_main.copy().drop(columns=['student_id'])
print(f'student_id를 제거한 후 데이터프레임:\n {df_main_copy.head()}')
```

```python
# final_score와 assignment_rate의 상관관계 추출
    # corr: 상관계수 계산 함수
    # 상관계수는 -1에서 1 사이의 값으로 나타나며, 1에 가까울수록 강한 양의 상관관계, -1에 가까울수록 강한 음의 상관관계를 의미
correlation_matrix = df_main_copy.corr()
print("\n 상관계수 행렬:\n", correlation_matrix)
```

```python
# loc: 특정 행과 열을 선택하는 함수
# 'final_score' 행과 'assignment_rate' 열의 상관계수 추출
final_score_assignment_rate_corr = correlation_matrix.loc['final_score', 'assignment_rate']
print("final_score와 assignment_rate의 상관관계: ", final_score_assignment_rate_corr)
```

## 2-4. 분석 결과

1. 산점도에서 점들이 전반적으로 **우상향하는 패턴**을 보이며, 이는 두 변수 사이에 **양의 상관관계**가 있음을 의미
2. `.corr()` 메서드로 계산한 상관 계수 역시 양수 값을 가지며, '과제 제출률'이 높을수록 '시험 점수'도 높은 경향이 있음을 수치적으로 뒷받침
